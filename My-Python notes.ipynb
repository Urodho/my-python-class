> Always remmber to indent the right way

Here are the guidelines by PEP 8 on indentation : https://peps.python.org/pep-0008/#indentation

```python
# Grade attained and Target grade
attained_grade = 60
target_grade = 50

# Comparing the grades

if attained_grade >= target_grade:
    print("Target Grade Achieved")
# Check if they haven't hit the target grade    
elif attained_grade < target_grade:
    print("Target Grade Not Achieved")
# Check if the grade is equal to zero
elif attained_grade == 0:
    print("You have a grade of zero, Please contact your teacher for assistance")    
```

```python
# Target Sales and Actual Sales
target_sales = 1000
actual_sales = 3000

# Compare our sales
if actual_sales >= target_sales:
    print("Target Sales Achieved")
# Check if we are close to hitting the target sales
elif actual_sales >= 920 :
    print("Target Sales Achieved")
else:
    print("Target Sales Not Achieved")    
    
```

You can have as many elif as u like, but you can only have one if and one else

|Keyword |Function | Use |
|------|--------|----------|
|"if" | check if a condition is met | First in the work flow |
|"elif" | Else check if another condition is met | After "if"|
| "else" | Else peform this action | After "elif" or "if" |

Accessing Elements in a List

```python
# Indexing in Python starts at 0, so the first item in the list is at index 0, the second item is at index 1, and so on.

grades[0] # This will return the first item in the list, which is 10
grades[2] # This will return the third item in the list, which is 40
grades[3] # This will return the fourth item in the list, which is 50
grades[-1] # This will return the last item in the list, which is 50
```
